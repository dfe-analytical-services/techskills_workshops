## Aggregate functions

The aggregates generate quick sumaries of data, performing a calculation that returns a single value.

The most commonly used are;
- `MIN()` - returns the smallest value in a column
- `MAX()` - returns the largest value in a column
- `COUNT()` - returns the number of values in a column
- `SUM()` - returns the sum of a column
- `AVG()` - returns the average of a column

### Quick note on NULL values

An important feature of the aggrgate functions is that they will **ignore any `NULL` values**. A `NULL` value is nether a zero (`0`) nor an empty string (` `). It is a value that has been left blank upon record creation.

The exception to this is counting the total number of rows in a table (`COUNT(*)`), which will inlude `NULL` values.

You can test for `NULL` values using `IS NULL` with the following syntax;
```
SELECT column FROM table WHERE column IS NULL
```
Inversely, `IS NOT NULL` will check for values that are not `NULL`.

## GROUP BY

Often when summarising data, the same summary may need to be applied to a group of data. For example finding the number of unique games in each genre.

For this, add the `GROUP BY` operation to the end of the SQL query. For example;
```
SELECT genre, COUNT(DISTINCT game_name)
FROM catalog_40_copper_analyst_training.steam.games_ranking
GROUP BY genre;
```
This will create a table with a genre column, and a column named "COUNT(DISTINCTgame_names)" with a count for each genre.

To rename the aggregate column to something neater, you can use the `AS` followed by the new name. If this name has spaces, you must enclose it in backticks ``. For example;
```
SELECT genre, COUNT(DISTINCT game_name) AS `Number of games`
FROM catalog_40_copper_analyst_training.steam.games_ranking
GROUP BY genre;
```

## INSERT

`INSERT` is used to add a new record (row) into existing tables of data.

For example;
 
```
INSERT INTO table (column_1_name, column_2_name, ...) VALUES (value_1, value_2, ...)
```

Column details do not need to be specified. If left blank, the values will be matched to columns by the order they appear. This means the first value listed will be added under the leftmost column.

Missing details for columns not specificed in the values list will be insered as `NULL`.

If you want to insert multiple rows, you can sepearye lists of values with commas.

For example;

```
INSET INTO catalog_40_copper_analyst_training.steam.games_ranking (game_name, genre, rank_type, rank)
VALUES
('Portal 2', 'Adventure', 'Review', 13),
('DELTARUNE', 'Adventure', 'Review', 14),
('Servant of the Lake', 'Adventure', 'Review', 15);
```

### Data types

When insterting data into a column, the value must be of the same data type as the column. For example an integer value must be provided for the `rank` column, as this is an intiger column. You can check the type of data in a column by using the `TYPEOF()` function.

For example;

```
SELECT TYPEOF(rank) FROM catalog_40_copper_analyst_training.steam.games_ranking;
```
