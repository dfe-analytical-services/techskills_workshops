# Joining data

Often joing two or more tables expands the insights we can draw from a signle table on its own.

This is reliant on a common column that can be across the tables to be joined.

## Types of join

It is rare to have data sets where the data in the common column is idential across the tables. As such we may need to specify what to do with values that don't appear in one table, but do in the other, or values that appear in the column multiple times.

This is done by specifying the type of join. These are;
- `INNER JOIN` - an inner join where only matched data is retained in the joined table.
- `LEFT JOIN` - an outer join where all data in the first table is retained in the joined table, with matched data from the second table.
- `RIGHT JOIN` - an outer join where all data in the second table is retained in the joined table, with matched data from the first table.
- `FULL JOIN` - an outer join where all data in both tables is retained.
