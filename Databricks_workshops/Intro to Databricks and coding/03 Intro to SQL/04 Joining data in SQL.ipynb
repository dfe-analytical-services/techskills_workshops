# Joining data

Often joing two or more tables leads to greater insights into the data than can be drawn from a single table on its own.

However, joining tables is reliant on a common column that exists in the tables to be joined, for example, a unique reference number.

## Types of join

It is rare to have data sets where the data in the common column is idential between the tables. As such we may need to specify what to do with values that don't appear in one table, but do in the other.

This is done by specifying the type of join. These are;
- `INNER JOIN` - an inner join where only matched data is retained in the joined table.
- `LEFT JOIN` - an outer join where all data in the first table is retained in the joined table, with matched data from the second table.
- `RIGHT JOIN` - an outer join where all data in the second table is retained in the joined table, with matched data from the first table.
- `FULL JOIN` - an outer join where all data in both tables is retained.

To note an LEFT JOIN of table a onto table b is the same as a RIGHT JOIN of table b onto table a.

An example join would look like;
```
/* Selects all of table a and all of table b, from table a, and joins on table b
the operation `ON` specifies which common columns are equivalent */

SELECT example_catalog.example_schema.table_a.*, example_catalog.example_schema.table_b.*
FROM example_catalog.example_schema.table_a
INNER JOIN example_catalog.example_schema.table_b
ON example_catalog.example_schema.table_a.column_name = example_catalog.example_schema.table_b.column_name;
```

### AS

When joining tables, you may need to refer to them multiple times in the SQL query, which is a lot of text, especially if the table names are long. To make the code look neater you can use `AS` to provide an alias for the table name to use futher on in your code.

For example;
```
SELECT a.*, b.*
FROM example_catalog.example_schema.table_a AS a
INNER JOIN example_catalog.example_schema.table_b AS b
ON a.column_name=b.column_name;
```

Is equivalent to the earlier `INNER JOIN`, but uses far cleaner table names.

## Duplicate data

There may be instances where there are duplicates in the common column between tables (e.g. the same unique reference numbers appear in multiple rows in table_a).

This can create some many-to-one, one-to-many and many-to-many joins, where rows of data are repeated across a table for each entry that matches to the other table.

## UNION

`UNION` will join the results of two `SELECT` tables, by appending the data onto the first `SELECT`.

The syntax is a `SELECT` query followed by `UNION` followed by another `SELECT` query. This can be repeated as many times as you have tables to join.

For this to work, you will need to have the same number of columns in each `SELECT`, and the columns must be able to accept similar data types.

## UNION ALL

Hopefully, it has been demonstrated how `UNION` will automaticall drop any duplicate rows. For example there was only one row for Azlan, 28 in the joined table, even though they appear in both **table_a** and **table_c**.

If you are wanting duplicates to appear, use `UNION ALL` instead.

Try it out; replace `UNION` with `UNION ALL` in your previous join. You should now have a table with 7 rows (and 3 Brians ! )
