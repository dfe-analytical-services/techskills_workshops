# Creating your own table

For this section, we will be adding, removing, and updating data in tables. To avoid changing data future colleagues may use in this training, let's create our own table to manipulate.

## CREATE TABLE

The syntax to create a table starts `CREATE TABLE IF NOT EXISTS` this tells SQL to create a new table if one with that name doesn't already exist. Then you should specify where the table will be stored, and the table's name in the following structure;

**name_of_catalog.name_of_schema.name_of_table**

If I wanted to save a table named **my_table** in the **steam** schema, in the **catalog_40_copper_analyst_training** catalog. I would specifcy `CREATE TABLE IF NOT EXISTS catalog_40_copper_analyst_training.steam.my_table`.

Then, within brackets, you should list the columns in your new table, with the format `column_name data_type constraint`, where data type can be;
- VARCHAR(n) - a String/Character variable of length up to n
- CHAR(n) - a String/Character variable of fixed length n
- INTEGER - an Integer
- SERIAL - an Incrementing Integer for example a key column
- DATE - a Date stored in the form YYYY-MM-DD
- DATETIME - a Date and Time in the form YYYY-MM-DD HH:MI:SS

And where constraint is an optional element where you can add constraints on the data that can be entered into a column.

For example,
```
CREATE TABLE IF NOT EXISTS example_catalog.example_schema.example_table (
    ID INTEGER PRIMARY KEY,
    name VARCHAR(255),
    age INTEGER CHECK (age < 150),
    email VARCHAR(255) UNIQUE
);
```

Creates a table called example_table, that has 4 columns, ID, name, age, and email. ID and age and INTEGER columns, and name and email are String/Character columns where strings cannot exceed 255 characters. ID has the contraint of being the PRIMARY KEY, age has the CHECK contraint, requiring to be under 150, and email has the UNIQUE contraint requiring all entries into this column to be unique (no two rows can share an email address).

## INSERT

`INSERT` is used to add a new record (row) into existing tables of data.

For example;
 
```
INSERT INTO table (column_1_name, column_2_name, ...) VALUES (value_1, value_2, ...)
```

Column details do not need to be specified. If left blank, the values will be matched to columns by the order they appear. This means the first value listed will be added under the leftmost column.

Missing details for columns not specificed in the values list will be insered as `NULL`.

If you want to insert multiple rows, you can sepearye lists of values with commas.

For example;

```
INSET INTO catalog_40_copper_analyst_training.steam.games_ranking (game_name, genre, rank_type, rank)
VALUES
('Portal 2', 'Adventure', 'Review', 13),
('DELTARUNE', 'Adventure', 'Review', 14),
('Servant of the Lake', 'Adventure', 'Review', 15);
```

### Data types

When insterting data into a column, the value must be of the same data type as the column. For example an integer value must be provided for the `rank` column, as this is an intiger column. You can check the type of data in a column by using the `TYPEOF()` function.

For example;

```
SELECT TYPEOF(rank) FROM catalog_40_copper_analyst_training.steam.games_ranking;
```
