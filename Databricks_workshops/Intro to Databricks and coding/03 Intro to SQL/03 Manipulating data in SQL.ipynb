# Creating your own table

For this section, we will be adding, removing, and updating data in tables. To avoid changing data future colleagues may use in this training, let's create our own table to manipulate.

## CREATE TABLE

The syntax to create a table starts `CREATE TABLE IF NOT EXISTS` this tells SQL to create a new table if one with that name doesn't already exist. Then you should specify where the table will be stored, and the table's name in the following structure;

**name_of_catalog.name_of_schema.name_of_table**

If I wanted to save a table named **my_table** in the **steam** schema, in the **catalog_40_copper_analyst_training** catalog. I would specifcy `CREATE TABLE IF NOT EXISTS catalog_40_copper_analyst_training.steam.my_table`.

Then, within brackets, you should list the columns in your new table, with the format `column_name data_type constraint`, where data type can be;
- VARCHAR(n) - a String/Character variable of length up to n
- CHAR(n) - a String/Character variable of fixed length n
- INTEGER - an Integer
- SERIAL - an Incrementing Integer for example a key column
- DATE - a Date stored in the form YYYY-MM-DD
- DATETIME - a Date and Time in the form YYYY-MM-DD HH:MI:SS

And where constraint is an optional element where you can add constraints on the data that can be entered into a column.

For example,
```
CREATE TABLE IF NOT EXISTS example_catalog.example_schema.example_table (
    ID INTEGER PRIMARY KEY,
    name VARCHAR(6),
    age INTEGER CHECK (age < 150),
    email VARCHAR(100) UNIQUE
);
```

Creates a table called example_table, that has 4 columns, ID, name, age, and email. ID and age and INTEGER columns, and name and email are String/Character columns where strings cannot exceed 255 characters. ID has the contraint of being the PRIMARY KEY, age has the CHECK contraint, requiring to be under 150, and email has the UNIQUE contraint requiring all entries into this column to be unique (no two rows can share an email address).

## INSERT

`INSERT` is used to add a new record (row) into existing tables of data.

For example;
 
```
INSERT INTO table (column_1_name, column_2_name, ...) VALUES (value_1, value_2, ...)
```

Column details do not need to be specified. If left blank, the values will be matched to columns by the order they appear. This means the first value listed will be added under the leftmost column.

Missing details for columns not specificed in the values list will be insered as `NULL`.

If you want to insert multiple rows, you can separate lists of values with commas.

For example;

```
INSERT INTO example_catalog.example_schema.example_table (ID, name, age, email)
VALUES
(1, 'Audrey', 29, 'audrey@email.com'),
(2, 'Brenda', 45, 'brenda@email.com'),
(3, 'Celine', 32, 'celine@email.com');
```

### Data types

When insterting data into a column, the value must be of the same data type as the column. We have set this when  creating the table in `CREATE TABLE`. However, if you are not sure what the type of data is, becuase you are adding data to an existing column, you can check by using the `TYPEOF()` function.

For example;

```
SELECT DISTINCT TYPEOF(column_name) FROM table_name;
```

Running this will retrieve the data type for all rows in the table.

## ALTER

Once you have created a table, and now have data in your table, you may need to change the table, for example;
- Adding a column,
- Dropping a column,
- Renaming a column,
- Modifying the data type or contraints of a column,
- Renaming a table

These follow a similar syntax: `ALTER TABLE table_name SQL OPERATION column_name`.

For example;
```
ALTER TABLE example_catalog.example_schema.example_table
ADD COLUMN contact_number VARCHAR(11);
```

Will add a new column **contact_number** (the new column will autopopulate with `NULL` values for all rows of data)

And;
```
ALTER TABLE example_catalog.example_schema.example_table
DROP COLUMN age;
```

Will drop the existing column **age**

- **To note**: `DROP COLUMN` may not be supported for your Delata table. To work around this you should first enable column mapping on the table. You can do this by running `ALTER TABLE example_catalog.example_schema.example_table SET TBLPROPERTIES ('delta.columnMapping.mode' = 'name')` before the `DROP COLUMN` command.

And;
```
ALTER TABLE example_catalog.example_schema.example_table
RENAME COLUMN email TO contact_email;
```

Will rename the existing column **email** to **contact_email**.

And;
```
ALTER TABLE example_catalog.example_schema.example_table
ALTER COLUMN name TYPE VARCHAR(15);

ALTER TABLE example_catalog.example_schema.example_table
ALTER COLUMN name SET NOT NULL;
```

Will modify the data `TYPE` of **name** to allow character strings up to 15 characters long, and `SET`s the constraint to `NOT NULL`, meaning the all rows must have a name.

- **To note**: The Delta lake will not allow you to narrow column types (e.g. reducing the maximum length allowable) due to the potential for data loss. Instead you should create a new table with the desired string length defined.

We can do that with `UPDATE`

The syntax for `UPDATE` is `UPDATE table_name SET column_1_name = value_1, column_2_name = value_2 WHERE condition;`

It is very important to remember the `WHERE` condition: if this is not specified all records will be updated.

## DELETE

Now we know how to inset and update rows of data, how do we remove rows? This is done with `DELETE`.

The syntax is `DELETE FROM table_name WHERE condition;`

Just like in `UPDATE`, it is important to specify the `WHERE` condition. Without it, **all data will be deleted**.

It is good practice to run a `SELECT` query with the `WHERE` condition before you `DELETE` to make sure you are specifying the correct rows of data.
