The `SELECT` operation is partnered with `FROM` to help direct the query to where it should retieve data from.

To query a full table, we would use
```

```
where the asterisk `*` denotes all columns.

## Bracketing

It may be that you want to select a table with multiple distinct columns, or with multiple columns where only some are unique. You can do this by adding round bracketsafter the `DISTINCT` keyword.

For example;
```
SELECT DISTINCT(game_name, rank_type), rank
FROM catalog_40_copper_analyst_training.steam.games_ranking;
```

Which will group game name and rank type, and only show unique groups of those two columns. Whereas rank will be treated seperately, and will allow duplicate values.

## WHERE

Another operation you can use with `SELECT` to be more specific is `WHERE`.

This will only select data that meets a specified condition.

To add contions with `WHERE` you will need to use logical operators, such as;
- `=` for equal to,
- `>` for greater than,
- `>=` for greater that or equal to,
- `<` for less than,
- `<=` for less than or equal to,
- `<>` for not equal to

If the condtion relates to a character sting (e.g. selecting all rows where the game_name collumn is equal to `'Terraria'`) you must enlcose the string in single quotation marks. For numeric conditions, (e.g. selecting all rows where the rank column is equal to `1`) the quotations are not needed.

## WHERE conditions

Beyond simple logical operations, you can also specify more complex conditions using operations including;
- `AND`, specify multiple conitions that must all be met,
```
SELECT * FROM tablename WHERE (column1 = 'Condition 1') AND (column2 <> 'Condition 2');
```
- `OR`, specify multiple conditions, where rows meeting at least 1 conditon are selected,
```
SELECT * FROM tablename WHERE (column1 = 'Condition 1') OR (column2 <> 'Condition 2');
```
- `IN`, selects rows where items in a coulm are in a list of alowed values,
```
SELECT * FROM tablename WHERE column1 IN ('Condition 1', 'Condition 2', 'Condition 3');
```
- `BETWEEN`, selects rows where items are between two numeric values,
```
SELECT * FROM tablename WHERE column1 BETWEEN 0 AND 100;
```

### LIKE

`LIKE` is used to select rows that match a pattern.

The pattern can be specificed using `_` and `%`.

- `_` inidcates a single character, so 'a_t' would match to 'ant' and 'act',
- `%` incicates multiple characters, so 'a%t` would match to 'ant' and 'act', and also 'accomplishment' and 'ambivalent'
   - `%` also include matching to zero characters, so 'a%t' would also match to 'at'

To note, the patterns in `LIKE` are case sensitive, so 'a_t' would **not match** to 'Ant'. Consider first using `LOWER` or `UPPER` to transform the column to all lower case or all upper case before pattern matching, if the case is not important.

### NOT

Adding `NOT` before some keywords (e.g. `NOT IN`, `NOT BETWEEN`, or `NOT LIKE`) will select the inverse, e.g. all values that do `NOT` appear in a list of allowed values.

`NOT` can also be used in conjuntion with `=`, `<`, and `>` conditions too, with the syntax `SELECT * FROM tablename WHERE NOT column = value;`

## EXISTS

`EXISTS`can also be used with `WHERE` in `SELECT` queries. This checks if a secondary query returns any rows

For example;
```
SELECT DISTINCT game_name
FROM catalog_40_copper_analyst_training.steam.games_ranking AS ranking
WHERE EXISTS (
    SELECT DISTINCT game_name
    FROM catalog_40_copper_analyst_training.steam.steam_game_reviews AS reviews
    WHERE recommendation = 'Not Recommended'
    AND reviews.game_name = ranking.game_name
);
```

Will select games names from the **games_ranking** table, where rows exist in the **steam_game_reviews** table filtered for games where the recommendation is "Not Recommended".

## ORDER BY

When selecting data, there is no guarantee that the data will be in any specific order.

However, you can use the `ORDER BY` operation to sort columns in ascending, `ASC`, or descending, `DESC` order. By default (when neither are specified), `ORDER BY` sorts in ascending order.

You can specify multiple columns to sort, and the sorts will occur in the oder of appearance. For example,

```
SELECT * FROM catalog_40_copper_analyst_training.steam.games_ranking
ORDER BY rank ASC, game_name ASC;
```

will first sort rank in ascending order. Then for each unique game name within each rank (all games with rank 1, all games with rank 2 and so on), will be sorted alphabetically.

## Aggregate functions

The aggregates generate quick sumaries of data, performing a calculation that returns a single value.

The most commonly used are;
- `MIN()` - returns the smallest value in a column
- `MAX()` - returns the largest value in a column
- `COUNT()` - returns the number of values in a column
- `SUM()` - returns the sum of a column
- `AVG()` - returns the average of a column

### Quick note on NULL values

An important feature of the aggrgate functions is that they will **ignore any `NULL` values**. A `NULL` value is nether a zero (`0`) nor an empty string (` `). It is a value that has been left blank upon record creation.

The exception to this is counting the total number of rows in a table (`COUNT(*)`), which will inlude `NULL` values.

You can test for `NULL` values using `IS NULL` with the following syntax;
```
SELECT column FROM table WHERE column IS NULL
```
Inversely, `IS NOT NULL` will check for values that are not `NULL`.

## CASE

Similar to the aggregation functions, `CASE` can be used with a `SELECT` query.

`CASE` is used alongside `WHEN`, `THEN`, and `ELSE` to apply a conditional column to the data. It works by checking a condition (`WHEN ...`), defines the value when the condition is met (`THEN ...`), and if no conditions are met sets a default (`ELSE ...`). If no default is set, any row that does not meet the conditions will be `NULL`.

The syntax is;
```
SELECT column1, column2,
CASE
    WHEN column2 = condition THEN 'Pass'
    WHEN column2 = condition THEN 'Fail'
    ELSE 'Unknown'
END AS column3
FROM table_name;
```

## GROUP BY

Often when summarising data, the same summary may need to be applied to a group of data. For example finding the number of unique games in each genre.

For this, add the `GROUP BY` operation to the end of the SQL query. For example;
```
SELECT genre, COUNT(DISTINCT game_name)
FROM catalog_40_copper_analyst_training.steam.games_ranking
GROUP BY genre;
```
This will create a table with a genre column, and a column named "COUNT(DISTINCTgame_names)" with a count for each genre.

### Aliases

Sometimes you may want to rename the column the aggregate function creates (as "COUNT(DISTINCTgame_names)" is not the neatest of names.

This can be done using the `AS` keyword to give the column name an alias.

For example;
```
SELECT genre, COUNT(DISTINCT game_name) AS `games_count`
FROM catalog_40_copper_analyst_training.steam.games_ranking
GROUP BY genre;
```

If your new name has spaces, you must enclose it in backticks `` (which can be found next to the "1" key on most keyborads). For example ```AS `Number of games` ```

Run the below code to explore it at work.
