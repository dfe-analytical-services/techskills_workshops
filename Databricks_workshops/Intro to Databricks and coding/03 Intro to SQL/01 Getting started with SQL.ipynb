# What is SQL?
SQL is a standard language for accessing and manipulating data stored in relational databases. It stands for Structured Query Language.

## What is a relational database?
A relational database is a digital database based on the relational model of data. This is where data is stored as a collection of tables, with each table containing a set of rows and columns.

Each row contains the information for a single observation - e.g. a child, a provider, a transaction - while each column contains a specific attribute for all rows in the data - e.g. a name, an id number, a transaction date.



## Quick note on syntax
Most SQL queries will follow a similar structure containing the SQL Operation, followed by the table names, and column names if required. 

**All queries should end with a semi-colon to run it.**

For example, it could look like
```
SQLOPERATION tablenames columnnames;
```

While SQL Operations are not case sensitive (they can be in either upper case or lower case), keeping them capitalised can make your query more readable.

## Writing comments
It is typically good practice to add comments alongside code to explain what it does. In SQL comments are identified with `/*`.

Adding `/*` will comment out all content following on from it, including code on new lines. To only comment out specific lines, you should add `*/` at the end of the comment.

If you are only commenting one line, you can use `--` at the beginning of the line, to comment out that one line only.
