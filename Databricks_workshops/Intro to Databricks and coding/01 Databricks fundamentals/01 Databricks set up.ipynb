## 1. Key Differences from Other Tools

If you're coming from R Studio, SPSS, or Python on your local machine, here are the main differences:

| Feature | Local tools (RStudio, SPSS) | Databricks |
| --- | --- | --- |
| **Data storage** | Files on your machine or network drive | Tables in Unity Catalog (cloud storage) |
| **Processing** | Limited to your machine's RAM/CPU | Distributed across a cluster (scales up) |
| **Collaboration** | Share files via email/SharePoint | Shared workspace, Git integration |
| **Languages** | Usually one per tool | Python, SQL, R, Scala in one notebook |
| **Version control** | Manual or separate Git setup | Built-in Git integration |
| **Data governance** | Manual access controls | Unity Catalog permissions and lineage |

### Other key differences
- **No local files** — data lives in cloud storage accessed via Unity Catalog, not on your C: drive
- **Compute is separate from storage** — you attach to a cluster to run code, but data persists independently
- **Notebooks are collaborative** — multiple people can view/edit (though simultaneous editing needs care)
- **Everything is auditable** — queries, data access, and changes are logged
- **Automation** - Complex pipelines can be set up using Databricks workflows and set to automatically run, either on a timer or a specific trigger

## 3. Workspaces

A **workspace** is your Databricks environment — it's where you write code, manage data, and run jobs.

### What's in a workspace?
- **User folders** — Each user has a personal folder (`/Users/your.email@education.gov.uk/`)
- **Shared folders** — Team or project folders that multiple users can access
- **Repos/Git folders** — Cloned repositories from Azure DevOps

### Workspace organisation at DfE
- Workspaces are typically separated by **environment** (dev, test, prod) or **team**
- Your personal folder is your sandbox — experiment freely here
- Production/shared work should live in Git-controlled repos
- Keep your workspace tidy: use folders, meaningful names, and archive old work

### Best practices
- **Don't store data in your workspace** — use Unity Catalog tables or Volumes
- **Use Git folders** for anything that needs version control or collaboration
- **Name notebooks clearly** — include numbers for ordering (e.g., `01_data_load`, `02_transform`)
- **Share via permissions**, not by copying files into other people's folders

## 4. Compute — Clusters and SQL Warehouses

In Databricks, "compute" is the processing power that runs your code. There are different types:


| Compute type | Purpose / Use cases | Key features / Notes | Configuration | Code language support | Availability |
| --- | --- | --- | --- | --- | --- |
| Serverless | Quick start to query in SQL or Python | Minimal warm-up time. No infrastructure to manage. No query can run longer than 48 hours | Handled by Databricks | SQL, Python | All workspace users |
| SQL Warehouse | Running SQL queries and powering dashboards. Best for analysts writing SQL, BI tools, scheduled queries. Only compute option for SQL Editor | Auto-scales, auto-suspends, optimised for SQL workloads. Shared resource so more efficient. Relatively short boot up time. Recommended compute for any SQL activities | Created by ADA team for each catalog | SQL | Only other users of same warehouse (typically your team) |
| Personal Cluster | Interactive development for a single user. Exploratory work in Python/R/Scala, ML experimentation. Only option for R coding. Use when you need specific versions or libraries | Single-user access, full library control. Install packages, choose runtime version. Shuts down after being unused for an hour. Least cost effective so only use if needed. Takes a few minutes to boot up. Compute power limited to standard or large but can be raised if necessary | Set up by you with choice of standard (roughly same power as DfE laptop) or large (roughly double standard) | SQL, Python, R, Scala | Only you |
| Shared Cluster | Shared resource to be more cost effective. Where using Python, R or Scala please use this or Serverless | Multiple users share resources, managed by admins. Less control over packages/config vs. personal clusters. Takes a few minutes to boot up | Created by ADA team for specific projects | SQL, Python, R, Scala | Other users on same project |


### Which should I use?
| Task | Recommended compute |
| --- | --- |
| Quick ad-hoc SQL or Python queries | Serverless |
| SQL queries, dashboards, and SQL Editor | SQL Warehouse* |
| Python or Scala development | Shared Cluster or Serverless |
| R coding or need specific libraries/versions | Personal Cluster or Shared Cluster |
| Team project work (Python/Scala) | Shared Cluster |
| Scheduled production jobs | Job Cluster (auto-created) |

*You should use SQL warehouse unless:
- You need to read in data from Databricks volumes
- There are likely to be multiple people running intensive processes in your team (who are on the same SQL warehouse) at the same time, to the point it would hit the processing limit
- You/your team don't have access to a SQL warehouse 


> **Cost note:** Clusters cost money while running. Always let them auto-terminate when not in use. SQL warehouses auto-suspend after a few minutes of inactivity.


Note that shared clusters and SQL warehouses have to be requested through the [ADA team](adaptteam@education.gov.uk).

In most cases a personal cluster will be the most versatile and easily accessible option. However your team may want to request a shared SQL warehouse if you have a lot of processing heavy SQL queries and do not need to use any other language in the Databricks platform itself.

All compute options can be used both within the Databricks platform and be connected to through other applications. Instructions on how to connect R / RStudio to a SQL Warehouse, or a personal cluster can be found on the following pages:

- [Setup Databricks SQL Warehouse with RStudio](https://dfe-analytical-services.github.io/analysts-guide/ADA/databricks_rstudio_sql_warehouse.html)
- [Setup Databricks Personal Compute cluster with RStudio](https://dfe-analytical-services.github.io/analysts-guide/ADA/databricks_rstudio_personal_cluster.html)

Once you have a compute resource you can begin using Databricks. You can do this either through connecting to Databricks through RStudio, or you can begin coding in the Databricks platforms using scripts, or Notebooks.

### ✅ Task: Set up a Personal Cluster

Follow these steps to create your own personal compute cluster:

1. In the left sidebar, click **Compute**
2. Click the **Create compute** button (top right)
3. Under **Policy**, select **DfE Personal Compute**
4. Give your cluster a name (e.g., `SURNAME, Firstname's DfE Personal Compute Cluster`)
5. Under **Performance**, choose your cluster size:
   - **Standard** — roughly the same power as a DfE laptop (suitable for most tasks)
   - **Large** — roughly double the power of standard (for heavier workloads)
6. Leave other settings as default unless you have a specific need
7. Click **Create compute**
8. Wait a few minutes for the cluster to start — you'll see the status change from "Pending" to "Running"

Databricks charges based on the usage of compute, so if you finish work on Databricks it is good practice to terminate it when you're done to save money for the Department. As a precaution against running up unnecessary charges your personal cluster is set to terminate after 1 hour of inactivity.

#### Things to note
- Your cluster will **automatically shut down** after 60 minutes of inactivity
- To restart it later, go to **Compute** and click the **Start** button next to your cluster
- You can **terminate** your cluster manually when you're finished to save costs
- If you need to install specific libraries, you can do so via the **Libraries** tab on your cluster's page

## 5. Unity Catalog, Schemas, and Tables

More information on Unity Catalog, Schemas and Tables can be found in the [Analyst Guide](https://dfe-analytical-services.github.io/analysts-guide/ADA/databricks_fundamentals.html#unity-catalog)

The ‘unity catalog’ is a single catalog that contains all the other catalogs of data in the Department. Catalogs are very similar in concept to a SQL database in that they they contain schemas, tables of data and views of data.

The majority of data and files on Databricks should be stored in the ‘unity catalog’. This is similar in concept to a traditional database server, however the unity catalog also contains file storage in the form of volumes.

The unity catalog can be accessed through the ‘Catalog’ option in the Databricks sidebar.

Unity Catalog organises all data in a **three-level namespace**:

```
Catalog > Schema (Database) > Table/View/Volume
```

### Structure

Like a SQL database a catalog has schemas, tables, and views which store data in a structured (usually tabular) format.

- **Catalog** — Top-level container (e.g., `catalog_name`). Often represents an environment or department.
- **Schema** A schema is a sub-division of a catalog which allows for logical separation of data stored in the catalog. — Groups related tables together (e.g., `schema_name`)
- **Table** — The actual data, equivalent to SQL tables and have the ability to turn on version control which audits each change to the data. referenced as `catalog.schema.table`
- **Views** — Saved SQL queries that act like tables but don't store data themselves; the query runs each time the view is referenced, allowing alternative formats without duplicating data
- **Volumes** — File storage similar to a shared drive, used for storing any type of file (CSVs, JSON, images, etc.) that can be accessed and manipulated through code



Tables and views sit within a schema and these are where you would store your core datasets and pick up data to analyse from.

Volumes are stored under a schema within a catalog. Files in here can be accessed and manipulated through code. You can also upload files to a volume through the user interface.

### Permissions
- Access is controlled at catalog, schema, or table level
- You can only see/query data you've been granted access to
- Use the **Catalog Explorer** (left sidebar > Catalog) to browse what's available to you

## 6. Cloning and Connecting to Azure DevOps

Databricks integrates with Azure DevOps (Git repos) for version control.

### How to clone a repo
1. In the left sidebar, click **Workspace** > navigate to your user Home folder
2. Click  **Create** > **Git folder**
3. Paste the **clone URL** from your Azure DevOps repo
4. Select the branch you want to work from and click **Create Git folder**

![Cloning Repo](../../../images/DatabricksWorkshop/Create_git_folder.png)

### Key points
- Each user clones their own copy of the repo into their workspace
- Changes are committed and pushed back to DevOps just like any other Git workflow
- You can create branches, pull, commit, and push from the Git dialog in Databricks
- `.gitignore` rules from your repo are respected
- Notebooks are saved as source files (`.py`, `.sql`, `.r`) — not proprietary formats

### Tips
- Always **pull** before starting work to get the latest changes
- Use meaningful commit messages
- Avoid working directly on `main` — create feature branches
