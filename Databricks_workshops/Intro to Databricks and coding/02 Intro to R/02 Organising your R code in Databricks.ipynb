### Other ways to assign variables

The `<-` arrow is the most common way to assign a value in R, and is the recommended style. However, there are a few other ways you might see:

| Syntax | Example | Notes |
| --- | --- | --- |
| `<-` | `x <- 10` | The standard and recommended way. Reads as "x gets 10" |
| `=` | `x = 10` | Works in most situations, but `<-` is preferred by convention |
| `->` | `10 -> x` | Right assignment: assigns the value on the left to the name on the right. Less common, but you may see it |

You are most likely to see `<-` and `=` in practice. The `->` right assignment is occasionally used but not recommended for everyday code. We will use `<-` throughout this workshop.

Try running the examples below to see that they all produce the same result:

````r
# Standard assignment with <-
a <- 5
print(a)

# Assignment with =
b = 10
print(b)

# Right assignment with ->
15 -> c
print(c)
````

## Task 3: _Use functions_

Now that you can create variables, let's look at **functions**. A function is a named piece of code that performs a specific task and (usually) returns a result. You have already used one as `print()` is a function!

Functions are used within R to perform calculations, check information and manipulate data. They are written as a name followed by brackets, with any inputs (called **arguments**) inside the brackets:

`function_name(argument1, argument2)`

Some functions take no arguments, some take one, and some take several. Let's look at a few examples using the variables you created earlier:

````r
# print() displays a value — it takes one argument
print(my_number)

# sqrt() calculates the square root — it also takes one argument
sqrt(my_number)

# paste() combines text — it can take multiple arguments
paste("My name is", my_name)
````

## Writing your own functions

So far we have used **built-in functions** like `print()` and `sqrt()`. These are functions that R provides for you. But one of the most powerful things about R is that you can also **write your own functions**.

A function is like a reusable recipe. You define it once, and then you can use it as many times as you like with different inputs. This is a key part of writing good, reusable code (and a core principle of RAP!).

Here is the basic structure of a function in R:

```
my_function <- function(input) {
  # Do something with the input
  result <- input * 2
  return(result)
}
```

Let's break this down:
- `my_function` is the name you are giving your function (just like naming a variable)
- `function(input)` says this is a function that takes one input (called a **parameter** or **argument**)
- The code inside `{ }` is what the function does
- `return(result)` sends the answer back

Try running the example below:

````r
# Define a function that doubles a number
double_it <- function(x) {
  result <- x * 2
  return(result)
}

# Use the function
print(double_it(5))   # Should print 10
print(double_it(20))  # Should print 40

# You can also use it with variables you created earlier
print(double_it(my_number))  # Should print 20 (because my_number is 10)
````

## Packages

So far we have used functions that are built into R (like `print()`, `sqrt()`, and `class()`). But one of R's great strengths is its **packages**. Packages are collections of extra functions that other people have written and shared.

Think of a package as a toolbox. R comes with a basic set of tools, but if you need something more specialised (like reading CSV files, cleaning data, or making charts), you install a package that contains those tools.

### What happens if you try to use a package you haven't loaded?

Let's see! Try running the cell below. It will produce an **error** — and that's the point. We want to see what the error looks like so you know how to recognise it in future:

````r
# This will produce an error because we haven't loaded the 'stringr' package yet!
str_to_upper("hello world")
````

You should see an error like: `could not find function "str_to_upper"`. This is R telling you it doesn't know about that function because the package it belongs to (`stringr`) hasn't been loaded.

To use functions from a package, there are **two steps**:

**Step 1: Install the package** (download it to your environment). You only need to do this once:

`install.packages("stringr")`

**Step 2: Load the package** (make its functions available in your session). You need to do this each time you start a new session:

`library(stringr)`

Notice that when we **install** a package, we put the package name in quotes (`"stringr"`). When we **load** the package with `library()`, we don't need quotes.

> **Note:** In Databricks, many common packages are already installed on the cluster, so you may only need `library()`. If you get an error saying the package is not available, then you need `install.packages()` first.

### The `package::function()` shortcut

Instead of loading a whole package with `library()`, you can also call a single function from a package using the `package::function()` syntax. For example:

`readr::read_csv("my_file.csv")`

This is useful when you only need one function from a package, or when two packages have functions with the same name and you want to be explicit about which one you mean.

Now let's fix our error from above. Run the cell below to install and load `stringr`, then use the function:

````r
# Step 1: Install the package (you may see a message saying it's already installed — that's fine)
install.packages("stringr")

# Step 2: Load the package
library(stringr)

# Now the function works!
str_to_upper("hello world")
````

## Comments and structure

Comments in code are some of the most useful documentation you can use, and are crucial for RAP to help make your code clear and reproducible. Comments are not run as code — they are used to explain what your code is doing.

You should start your script with comments that provide information such as what the code contains and what it is for. Consider what a new person would need to know in order to understand and run the code efficiently.

You can add comments in R using the `#` symbol. Any line that starts with a hash will be treated as a comment rather than code, and will not be run. For example:

````r
# This is a comment — R will ignore this line

my_number <- 10  # You can also add comments at the end of a line of code
````

Everything after the `#` on a line is treated as a comment, and everything before it is treated as code.

If you have multiple lines you wish to comment out, you can highlight them and use `Ctrl + /` to toggle comments on all the selected lines at once.

In Databricks notebooks, you can use markdown cells for longer explanations and headings, and use `#` comments within your code cells for shorter notes.

## Loading in the data

There are multiple ways to load data into R, including:
1. Reading a **CSV** file from a **Unity Catalog Volume** (cloud file storage)
2. Querying a **Unity Catalog table** directly (structured data, similar to a SQL database)
3. Reading a **local CSV** file from your computer (e.g. in RStudio)

In this notebook, we will focus on option 1, reading a CSV file stored in a Unity Catalog Volume.

### What is a Volume?

In Databricks, a **Volume** is a managed file storage location in the cloud. Think of it as a shared folder that your team can upload files to (CSVs, Excel files, etc.). It is **not** the same as:
- A **local file** on your computer (which you might use in RStudio with a file path like `"C:/Documents/data.csv"`)
- A **Unity Catalog table** (which is structured data you would query using SQL or `sparklyr`, not `read_csv()`)

Volume paths always start with `/Volumes/` followed by the catalog, schema, and volume name.

### Reading in CSVs from a Volume

We already loaded the `readr` package, which contains the `read_csv()` function for reading CSV files into R.

When reading data, we assign it to a named object using `<-` so we can reference it later:


chosen_name <- readr::read_csv("/Volumes/catalog_name/schema_name/volume_name/file.csv")


For this exercise, we will use `readr::read_csv()` to load the Airbnb Open Data dataset from the analyst training volume. This approach is simple and works seamlessly with tidyverse functions like `dplyr` and `tidyr`.

### Using SparkR for Large Datasets

For **very large datasets** that don't fit in memory, the package `SparkR`'s function `read.df()` can read files in a distributed way across the cluster:

## Cleaning data

Once you have loaded data into R, you will often need to clean it. The most common cleaning tasks are dealing with **missing values** and **duplicates**.

The `dplyr` and `tidyr` packages include useful functions for this, such as `filter()`, `mutate()`, `drop_na()` and `replace_na()`.

### Dealing with missing values

There are several approaches to handling NAs (missing values) in R:

| Approach | Function | What it does |
| --- | --- | --- |
| Remove all rows with any NA | `na.omit()` | Drops any row that contains a missing value in any column |
| Remove rows with NA in specific columns | `tidyr::drop_na(col1, col2)` | Drops rows only where the named columns are NA |
| Replace NAs with a value | `tidyr::replace_na()` | Substitutes NAs with a chosen value (e.g. 0, "x") |

### The pipe operator

Before we try these, note that R uses a **pipe operator** to chain operations together. It takes the output of the left-hand side and passes it as the first argument to the right-hand side. Think of it as saying "and then...".

There are **two** pipe symbols in R:

| Pipe | Source | Notes |
| --- | --- | --- |
| `%>%` | From the `magrittr`/`dplyr` packages (tidyverse) | The original pipe; requires loading a package |
| `\|>` | Built into base R (since R 4.1.0) | The native pipe; no package needed |

Both do the same thing in most cases. You will see `%>%` in older code and tidyverse tutorials, and `|>` in newer code. In this notebook we use `%>%` as it is still the most common in DfE projects, but either works.

For example:
```
# Using the tidyverse pipe
data %>% filter(age > 18) %>% arrange(name)

# Using the native pipe (equivalent)
data |> filter(age > 18) |> arrange(name)
```
Both read as: "take `data`, **then** filter for age > 18, **then** arrange by name."

## Task 8a: _In your own notebook, try cleaning the data of all rows with NAs, as below._

````r
# Task 8a: Remove ALL rows that contain any NA
# If every row has at least one NA, na.omit() will remove all rows.
# This can happen if the dataset has missing values in every row for at least one column.
airbnb_clean <- airbnb %>%
  na.omit()

# Print in the output how many rows were in the original dataset 
print(nrow(airbnb))

# Print in the output how many rows are in the cleaned dataset 
print(nrow(airbnb_clean)) 

# Calculate how many rows were removed after omitting NAs
rows_removed <- nrow(airbnb) - nrow(airbnb_clean)  # Subtract cleaned from original
print(rows_removed)
````


All 102,599 rows were removed from the Airbnb dataset when we dropped all rows containing any NA in any column. This happens because with 26 columns in a real-world dataset, every single row tends to have at least one missing value somewhere. This is a good example of why `na.omit()` can be too blunt an approach.

### Dealing with duplicates

The other common cleaning task is checking for and removing duplicate rows. The `distinct()` function from `dplyr` makes this straightforward.

- `distinct()` with no arguments removes rows that are completely identical across all columns
- `distinct(col1, col2, .keep_all = TRUE)` removes duplicates based on specific columns, keeping all other columns in the output

We will also show here how to label your print outputs so you can clearly see the results of each cleaning step.

## Task 9: _In your own notebook, use dplyr::distinct() to remove duplicate rows._ 

Try:

Remove completely duplicate rows from the Airbnb data. How many duplicates are there?
Remove duplicates based on room type and neighbourhood group. How many unique combinations are there?
Try a different column combination of your choice and compare the results


````r
# Task 9: Remove completely duplicated rows
# Use dplyr::distinct() to keep only unique rows
airbnb_unique <- airbnb %>%
  dplyr::distinct()

# Original
print(paste("Original row count:", nrow(airbnb)))   
# After removing duplicates  
print(paste("Row count after removing duplicates:", nrow(airbnb_unique)))

# Remove duplicates based on specific columns
# .keep_all = TRUE keeps all other columns in the result
# This is how you would get a distinct list of hosts by host name and neighbourhood group
airbnb_unique_hosts <- airbnb %>%
  dplyr::distinct(`host name`, `neighbourhood group`, .keep_all = TRUE)

print(paste("Row count after removing duplicates by host name and neighbourhood group:", nrow(airbnb_unique_hosts)))
````
