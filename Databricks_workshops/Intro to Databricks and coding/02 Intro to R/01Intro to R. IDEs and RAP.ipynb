## What is R? 

R is a coding language. There are many different languages of code, some others include SQL, python and JavaScript. They all have benefits, and the difference is often the syntax used (literally like learning a new language!). R is open-source, meaning that it is free, anyone can use it and anyone can contribute to developing new 'packages'. 

### What is a package?

A package is a set of functions that someone else has written and tied together in a neat bow, ready for you to use! You simply install the package, and then you have all the functions available. 

### What is a function? 

A function is a chunk of code that has been grouped together, given a name and often has 'placeholders' you can change, such that you can use that name to run the code, and apply it to different data. For example, mean(x) is a function that calculates the arithmetic mean of x. You replace x with any numeric data. 

## Using R with an IDE 

When writing code, you will usually work with something called an Integrated Development Environment (IDE). An IDE is a tool that makes coding easier by combining everything you need in one place, such as:

- A space to write code
- A console to run code
- Tools to view data and outputs
- Help documentation and file management. 

While it is technically possible to run R without an IDE, in practice almost everyone uses one because it makes coding much more efficient and user-friendly.

You can use any IDE you’re comfortable with to practise R, so feel free to download RStudio or Positron as instructed below to try them out, but in this guide we will be using Databricks notebooks.

## Choosing an IDE when working with R

There are many IDEs available when working with R.

At DfE, three commonly used IDEs are: 

### Databricks

At DfE, Databricks is part of the Analytical Data Access (ADA) service, which brings together departmental data and analytical tools in one place. It is different to some other IDEs, in that it is a cloud-based platform. It can be used as an IDE for writing and running R code using notebooks or standard code scripts. 

This is the environment used throughout this workshop and guide.

Databricks is particularly useful for working with large datasets. Once you have accessed your workspace and are working in a notebook, you will see the interface organised around notebooks and panels: 

- Main area - Notebook: Where you write and run the code in cells, with outputs shown directly underneath each cell. 

- Left sidebar - Workspace/Data/Files: Used to navigate notebooks, folders and datasets.

- Top bar - Notebook controls: Includes options to run cells, attach compute and manage the notebook. 

- Output area: Appears below each cell, showing results such as tables, charts and messages. 

### RStudio 

A widely used and beginner-friendly IDE designed specifically for R. 

It has useful areas and windows that enable you to see what tables you have loaded, charts you have created, Git user interface, file explorer, help windows and many more!

Normally you will see the screen split into 3 or 4 windows:


-   Top left - Source pane: Open and view your code scripts, tables, data sets, functions.

-   Top right - Environment/History/Connections/Git (if you have it): The environment shows you what data, functions, objects etc. you have.

-   Bottom left - Console: Shows what you have run, and you can type and run commands directly into the console.

-   Bottom right - File explorer/plots/packages/help/viewer.

#### How to get started with RStudio

Download R (language) and RStudio (IDE) from the DfE software centre. We also recommend that you download RTools (a helpful R extension) at the same time.


### Positron 

A newer IDE from Posit (the creators of RStudio). At DfE, analysts using R or Python outside of Databricks are encouraged to start exploring Positron, as it is where ongoing development from Posit and the wider open-source community is focused. 

It is designed to: 

- Support multiple languages (e.g. R and Python)

- Provide a more modern, flexible development experience

- Feel more similar to tools like VS Code. 

Positron is a good option if you prefer a more modern interface or plan to work across different languages. Normally you will see the screen arranged into panels, similar to other modern code editors: 

- Main area - Editor: Write and edit your scripts

- Bottom panel - Console/Terminal: Run R code and view output 

- Side bar - Files/Extensions/Environment: Navigate files, installed tools and project content 

- Outputs/plots panel: View charts, tables and other outputs.

## What is RAP? 

You may have heard the term RAP associated with R. RAP stands for Reproducible Analytical Pipeline. The full words still hide the true meaning behind the jargon though. What it actually means is using automation to our advantage when analysing data. It is as simple as using code such as an R script in your chosen IDE that we can click a button to execute and do the job for us.

Using R (the coding language) really helps us to put the 'R' in RAP (Reproducible). Ask yourself, if someone else picked up your work, could they easily reproduce your exact outputs? When the time comes around to update your analysis with new data, how easy is it for you to reproduce the analysis you need? In an ideal RAP, it would be as simple as plugging in the new data and clicking 'go', with no need to manually scroll through multiple scripts updating the year in every file name or every variable name that's changed. 

You can read more about RAP and DfE RAP principles on the [getting started with RAP](https://dfe-analytical-services.github.io/analysts-guide/RAP/getting-started-with-rap.html), [RAP for statistics](https://dfe-analytical-services.github.io/analysts-guide/RAP/rap-statistics/rap-statistics.html) and [RAP for general analysis](https://dfe-analytical-services.github.io/analysts-guide/RAP/rap-analysis/rap-analysis.html) pages on the Analysts' Guide.
