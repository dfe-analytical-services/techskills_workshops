## Working with notebooks and cells

Databricks notebooks are made up of cells. Each cell can contain either code (e.g., R code you want to run), or Markdown (text used for explanations, headings and notes) 

To create a new cell, click the '+' button in the notebook (above or below an exisiting cell). A new blank cell will appear where you can start typing. 

### Choosing a default language

Before you start, you should ensure your notebook is set to use R. At the top of the notebook, under the notebook name (most likely currently called 'New Notebook [date]), you will see the current language (e.g., SQL, Python or R). Once set, all code cells will run in the chosen language by default. You can override the default in each individual cell by clicking into it and choosing the language type from the cell toolbar.

### Markdown cells
Markdown cells are used to structure your notebook. For example, in a Markdown cell you can write: 
> 
> ## Load packages 
> This section loads the required packages 
> 

Use '#', '##' and '###' to create level 1, 2 and 3 headings. For example, above we use '## Load packages' to create a level 2 heading.


## Functions and packages

When using functions from packages, you can either type: 

`readr::read_csv()`

where `readr` is the package, and `read_csv()` is a function within the package, or just the function name on it's own. However, problems can arise if two packages contain functions with the same name. In this case R will use the function from the package loaded most recently. 

To avoid confusion, use `package::function()`. 

### Installing and using packages 

In Databricks, packages may already be available depending on your environment. If you need to install packages, you can do so in a cell. For example, to install `readr`, we would type: 

`install.packages("readr")`

To load the package ready to use, we would type: 

`library(readr)`

Notice that when we install a package, we put the package name in quotes. However, when we load the package, we don't need to. 

## Comments and structure

Comments in code are some of the most useful documentation you can use, and are crucial for RAP to help make your code clear and reproducible. Comments are not run as code, they are used to explain what your code is doing. 

You should start your script with comments that provide information such as what the code script contains and what it is for. consider what a new person would need to know in order to understand and run the code efficiently. 

You can add in comments in R using the # symbol. Any line that starts with a hash symbol will be treated as a comment rather than code, and so will not be 'run'. For example: 

```r
%r
# This is a comment 
```

You can also add comments to the end of lines of code by including the hash symbol half-way through a line. Everything after the hash will be treated as a comment, and everything before will be treated as code. 

If you have multiple lines of comments you wish to add, you can type them out, highlight them and then use `ctrl+/` to 'comment out' all of the highlighted lines.
