```sql
CREATE OR REPLACE TEMP VIEW schools_pupils_assessments 
AS 
SELECT pupil_id,
    school_urn,
    subject,
    score,
    grade,
    CASE 
        WHEN grade IN ('A*', 'A', 'B', 'C') THEN 1 
        ELSE 0 
    END AS a_star_to_c,
    assessment_date,
    term
FROM catalog_40_copper_analyst_training.bronze.assessments
LEFT JOIN (
    SELECT pupil_id,
        sex,
        dob,
        urn,
        year_group
    FROM catalog_40_copper_analyst_training.bronze.pupils_spring_2025
) AS pupils
USING(pupil_id) -- T-SQL: ON assessments.pupil_id = pupils.pupil_id
```
