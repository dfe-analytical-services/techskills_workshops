# What if you have both R and SQL code in your existing processes?

While we recommend Approach 1 for primarily R-based processes, and Approach 2 for SQL processes, it may be that your processes use a combination of these languages. If your workflow uses both R and SQL, you will likely need to use a blended approach to migration. You should weigh up your options and decide what is best for your process and team; this will depend on the context of your process, and your team skills and capacity.

For the below examples and exercises, we have created an example hybrid pipeline using both SQL and R in the **hybrid_pipeline** folder. You may recognise the process from the previous R-based and SQL-based processes.

## Move any SQL code into Databricks and run R code from RStudio / Positron

This approach is suitable when your existing SQL code is complex, well-tested or often reused and you want to keep it as SQL code. It involves running the SQL code directly in Databricks to create intermediate or final tables, which are then written to the Databricks catalog. The SQL processing happens entirely in Databricks, after which you connect to the Databricks catalog from RStudio, Positron or another IDE to read in the created tables and continue the process with R code.

An example of this approach has been included in the **hybrid_pipeline_solutions/databricks_example** folder to demonstrate how this might be done in practice.

## Translate any SQL code to R and run everything from RStudio / Positron

This approach is appropriate when your team primarily works in R and your SQL logic is not particularly complex or lengthy. It involves translating all existing SQL logic into R so that all your code is in the same language. The R code would then be run from RStudio / Positron / another IDE and would connect to the Databricks catalog to access the data as in Approach 1 above* (indeed, the solution to converting the hybrid pipeline using this approach is the same result as the _r_pipeline_ solution).

## Use an R wrapper function for SQL code and run everything from RStudio / Positron

This approach is a useful short-term or transitional approach when you want to reuse existing SQL code with minimal changes. It keeps SQL and R closely linked by embedding SQL queries within an R workflow. Any SQL code would first need updating from T-SQL to Spark SQL, where it could then be passed via R code using wrapper functions to run in Databricks, whilst R controls execution. You can run SQL from R by creating a reusable wrapper function that uses a Databricks connection (e.g., via the DBI and odbc packages) and executes queries with a function like dbGetQuery().

An example of setting up the connection and running queries can be found on the Databricks SQL Warehouse with RStudio page of the [Analysts Guide](https://dfe-analytical-services.github.io/analysts-guide/ADA/databricks_rstudio_sql_warehouse.html#pulling-data-into-rstudio-from-databricks). This could then be wrapped in a reusable function to standardise connection handling and query execution.

### Exercise: Migrate a combined process using R

Instead of running SQL code from within databricks, we can use functions like dbGetQuery and dbExecute to run SQL code directly from your R IDE.

Using the example process in the **hybrid_pipeline** folder, your task is to use functions from the odbc package to embed SQL code into the R code to allow the data to be read in from Databricks and the entire process to be completed from RStudio/Positron. You will need to translate the T-SQL code to run in databricks, however if you have already completed the previous exercises you should have this code to hand.

Using the translated SQL code, you will need to create an R pipeline that establishes a connection to databricks, executes SQL commands using dbExecute and reads SQL outputs using dbGetQuery, before performing the final analysis in R, and saving the output to your teams databricks catalog in a schema named 'your_name'. Remember to delete any code, files or folders that are no longer needed to maintain best practice.

This page on the [Analysts guide](https://dfe-analytical-services.github.io/analysts-guide/ADA/databricks_rstudio_sql_warehouse.html) contains some details that you may find helpful.  

When you have finished, re-run your updated pipeline and check you get no errors and the same outputs.

_For the solution, see **hybrid_pipeline_solutions/1_sql_wrapper_solution**._

If you have successfully managed to run SQL code chunks in R, you can also explore the get_clean_sql() function from the dfeR package, which can be used to read and reformat entire SQL files to a string that is executed at once. This approach does require SQL code to follow specific formats to work with the functions from odbc, however.

## Move all code into Databricks

This approach is best suited to processes you plan to automate using Databricks workflows or where keeping all processing in a single environment improves performance and reproducibility. It involves migrating all of your SQL and R code into the Databricks platform, executing it using notebooks or jobs.


### Exercise 2: Migrate a combined process using Databricks

Again, using the example SQL and R process in the **hybrid_pipeline** folder as a template, your task is to create three notebooks that run the pipeline as a job in Databricks.

The first notebook should contain the translated SQL code, including writing the intermediate output to a CSV in a Volume so that it can be used in the R section (hint - this requires a little python).

The second notebook should contain the R code, which reads the intermediate CSV from the Volume, performs the final data manipulation, and saves the final CSV in the same Volume.

The third notebook should contain the instructions to set up and execute the job that runs the pipeline. Detailed instructions for how to set up this notebook can be found on the [Analysts Guide](https://dfe-analytical-services.github.io/analysts-guide/ADA/databricks_workflows.html#scripted-workflows)

In order to run both SQL and R cells, you will need to create a Personal Compute Cluster by navigating to the Compute section form the left sidebar, clicking the blue 'Create with DfE Personal Compute' button, and selecting the 'DfE Personal R-Studio Compute' dropdown under the 'Policy' option.
