## Purpose
In Databricks you do not have to use a *Notebook* instead you can opt to use the more standard single *script* files you have previously been used to.

## Creating a file
Navigate to workspace in the Databricks sidebar:

![using_standard_code_scripts_1.png](../../../images/using_standard_code_scripts_1.png "using_standard_code_scripts_1.png")

Then find the folder in your workspace where you want to create a file.

![using_standard_code_scripts_2.png](../../../images/using_standard_code_scripts_2.png "using_standard_code_scripts_2.png")

Then click create file, Databricks will naturally default to a python script.

![using_standard_code_scripts_3.png](../../../images/using_standard_code_scripts_3.png "using_standard_code_scripts_3.png")

## Altering file type
There are two ways to change which language a script is written in.

1. You can rename the extension, these can be changed to any file type.

![using_standard_code_scripts_6.png](../../../images/using_standard_code_scripts_6.png "using_standard_code_scripts_6.png")

2. You can choose from the dropdown between python and SQL.

![using_standard_code_scripts_7.png](../../../images/using_standard_code_scripts_7.png "using_standard_code_scripts_7.png")

## Running the script directly

As with Notebooks you need to attach a compute instance to run a file.

- For python and SQL, any compute instance should be able to run python and SQL.
- R you will need to attach a dedicated personal compute set up to run R.
- Scala can also be added to a personal compute, bash scripts will also run, so if you are careful most things can be done.

If you click the play symbol at the top, the file will run and a window with output, terminal and debugger should appear.

![using_standard_code_scripts_8.png](../../../images/using_standard_code_scripts_8.png "using_standard_code_scripts_8.png")
