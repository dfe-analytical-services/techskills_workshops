# Connecting to repos

Just like any other way that you’ve worked with Git before, the first step is going to be to **clone your repo** inside Databricks. Git folders (previously called Repos in Databricks) can now be created in your Home folder.

In the blue menu on the left, click Workspace, and click the Home folder. The menu should look like this:

![Screenshot](/Workspace/Users/veera.madaan@education.gov.uk/connecting_to_repos.png)


Click the blue “Create” button in the top right corner, and then click “Git folder”:

![Screenshot](/Workspace/Users/veera.madaan@education.gov.uk/clone_git_repo.png)


Populate the field the box that comes up with the following information:

![Screenshot](/Workspace/Users/veera.madaan@education.gov.uk/create_git_folder.png)

- **Git repository url:** Copy the link to your repo (either in Azure DevOps or Github)
- **Git provider:** Choose your provider depending on where the repo resides
- **Git folder name**: This should automatically populate when you enter the url

Click “Create Git folder”. 

When the folder is created, you will be able to see it under your name in the Home folder.

**NB:** Databricks cannot clone very large repos. Sparse checkout should be selected in this instance to clone a selection of items in your repo. You can tell Databricks which elements of the repo you would like it to clone by specifying something called Cone Patterns. If you do not specify any Cone Patterns, then the Databricks default is to clone only files in the root folder, but none from any subdirectories. To specify your own Cone Patterns, enter them in the box provided. You can enter multiple patterns but if the folders that they refer to contain more than 800MB of data then your clone will continue to fail.
