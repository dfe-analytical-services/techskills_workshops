# Set up a connection to GitHub

### Getting Set up on GitHub

In this case, we will generate an access token in GitHub and give it to Databricks. To generate your GitHub access token, go to Github and click the user settings icon to the left of your initials in the top right of your screen - it looks like a person with a cog next to them:

### Step 1: Generate your access token  
In Github, create a fine-grained Personal Access Token with access to the required repositories and permissions by going to:

**User icon → Settings → Credentials → Fine-grained personal access token**

![Screenshot](/Workspace/Users/veera.madaan@education.gov.uk/git_access.png)

%md
Now that you have your access token, you should go straight to Databricks. 

### Step 2: Add your Git credentials
In the top right corner of the Databricks window, click your username and then:

**Settings → Linked accounts**

You will see the following screen:

![Screenshot](/Workspace/Users/veera.madaan@education.gov.uk/linked_accounts.png)

Click **Add Git credential**

You will see the following screen:
![Screenshot](/Workspace/Users/veera.madaan@education.gov.uk/github_credentials.png)

### Step 3: Choose Github
* **Git provider:** Set as 'Azure DevOps (Personal access Token)'

* Select **Personal access token** and paste your copied access token into the “Token” field at the bottom of the page. 

    **NB:** it is recommended to do this immediately to avoid losing it 

Click 'Save' at the bottom of the page, and now your connection between GitHub and Databricks is established
