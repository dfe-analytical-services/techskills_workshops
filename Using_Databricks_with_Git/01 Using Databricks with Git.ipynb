## 2. Why should I use Git with Databricks?

Using Git with Databricks gives you a structured way to track your code changes, collaborate with others and manage your code throughout its development.

### Key benefits
* Track changes over time instead of relying on memory or manual file copies
* Collaborate with others using branches and pull requests
* Roll back to earlier versions if something breaks
* Keep notebooks, Python files, SQL files, and deployment configuration in one versioned place
* Support a smoother path from interactive development to CI/CD

When you’re working on notebooks in Databricks without a Git connection, they are saved in your own personal Workspace, meaning that they are only accessible by you, unless you share them. If the notebooks are stored in a Github or Azure DevOps repo, your team can all access them by cloning the repo in Databricks.


%md
## 3. GitHub or Azure DevOps?

In Databricks, both can work well. The choice often depends on the tooling your organisation already uses.

* Choose **Azure DevOps** when your team already uses Azure Boards, Azure Pipelines, and Microsoft-native identity patterns.
* Choose **GitHub** when your team is already centered on GitHub repos, pull requests, and GitHub Actions.
* Choose **both** when you need internal enterprise repos in Azure DevOps and selected public or open-source collaboration in GitHub.

💡 Try to minimise unnecessary split workflows. If the team does use both, be explicit about what lives where and why.
