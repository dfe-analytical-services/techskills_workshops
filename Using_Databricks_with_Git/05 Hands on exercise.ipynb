# Suggested exercise

### Exercise goal
Connect a Git provider, clone a repository, make a small change, and push it on a feature branch.

### Exercise outline
1. Confirm Git folders are enabled
2. Add an Azure DevOps credential in Linked accounts
3. Clone a training repository
4. Create a branch named something like `feature/git-onboarding-yourname`
5. Edit a markdown cell in a notebook or update a README
6. Commit and push the change
7. Open a pull request in the Git provider

### Reflections
* What part of the flow felt most different from editing directly in the workspace?
* When would you use Azure DevOps versus GitHub in your team?
* What risks are reduced by working through Git rather than editing ad hoc?
