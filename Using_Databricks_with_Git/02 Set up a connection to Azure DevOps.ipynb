# Set up a connection to Azure DevOps

### Getting Set up on Azure DevOps

Access tokens are long strings of
 numbers and letters that act like a password between two services. They identify the user and their permissions from one service to another.

In this case, we will generate an access token in DevOps and give it to Databricks. To generate your Azure DevOps access token, go to DevOps and click the user settings icon to the left of your initials in the top right of your screen - it looks like a person with a cog next to them:

### Step 1: Generate your access token  
In DevOps, go to:

**User settings icon (icon resembles person with a cog) → Personal Access Tokens**

![Screenshot](/Workspace/Users/veera.madaan@education.gov.uk/devops-user-settings.png)

On the screen that appears, click the blue “New Token” button in the top right:

![Screenshot](/Workspace/Users/veera.madaan@education.gov.uk/devops-new-token.png)

The following window should appear:

![Screenshot](/Workspace/Users/veera.madaan@education.gov.uk/devops-create-token.png)
* Give the token a sensible name so that you can identify it in your list of tokens.
* The organisation field should be pre-populated for you, this can be left alone.
* You can modify the expiration date. Every time the token expires, you will have to follow this guidance again to set up a new token and reconnect to Git, so choose something sensible.
* You should make sure that “Full access” is selected under the Scopes heading.
When you have entered all of the required information, click “Create” at the bottom of the window.

A “Success!” window will open containing your new access token. 

**NB:** You must copy the token immediately, as you will not be able to access it again once this window is closed. If you fail to copy the token, you will have to regenerate a new one.

Now that you have your access token, you should go straight to Databricks. 

### Step 2: Add your Git credentials
In the top right corner of the Databricks window, click your username and then:

**Settings → Linked accounts**

You will see the following screen:

![Screenshot](/Workspace/Users/veera.madaan@education.gov.uk/linked_accounts.png)

Click **Add Git credential**

You will see the following screen:

![Screenshot](/Workspace/Users/veera.madaan@education.gov.uk/git_credential.png)

### Step 3: Choose Azure DevOps Services
* **Git provider:** Set as 'Azure DevOps (Personal access Token)'

* Paste your copied access token into the “Token” field at the bottom of the page. 

    **NB:** it is recommended to do this immediately to avoid losing it 
* Enter your email into the “Git provider username or email” field used to sign in to the Azure DevOps org

Click 'Save' at the bottom of the page, and now your connection between Azure DevOps and Databricks is established
